```python
#Task 2: Revenue & Product Performance Analysis
# Revenue by Product Type
product_revenue = (
    df.groupby("Product type")["Revenue generated"].sum().reset_index()
)
print("--- Total Revenue by Product Type ---")
print(product_revenue)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=product_revenue,
    x="Product type",
    y="Revenue generated",
    palette="viridis",
    hue="Product type",
    legend=False,
)
plt.title("Total Revenue Generated by Product Type")
plt.ylabel("Revenue")
plt.show()
plt.savefig("revenue_by_product_type.png")
plt.close()
```

```python
#Task 3: Inventory & Lead Time Analysis
# Stock levels vs Availability
print("--- Stock Level Statistics ---")
print(df[["Stock levels", "Availability", "Lead times"]].describe())


plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Stock levels", y="Availability", hue="Product type")
plt.title("Stock Levels vs. Availability")
plt.show()
plt.savefig("stock_vs_availability.png")
plt.close()
print(
    "\n[Success] Inventory analysis chart saved as 'stock_vs_availability.png'."
)
```

```python
#Task 4: Logistics & Carrier Performance Analysis
carrier_perf = (
    df.groupby("Shipping carriers")
    .agg(
        {
            "Shipping times": "mean",
            "Shipping costs": "mean",
            "SKU": "count",
        }
    )
    .rename(columns={"SKU": "Shipment count"})
    .reset_index()
)

print("--- Shipping Carrier Performance ---")
print(carrier_perf)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=carrier_perf,
    x="Shipping carriers",
    y="Shipping costs",
    palette="magma",
    hue="Shipping carriers",
    legend=False,
)
plt.title("Average Shipping Costs by Carrier")
plt.show()
plt.savefig("carrier_shipping_costs.png")
plt.close()
print(
    "\n[Success] Carrier performance metrics printed and chart saved as 'carrier_shipping_costs.png'."
)
```

```python
#Task 5: Manufacturing & Defect Rate Analysis
supplier_defects = (
    df.groupby("Supplier name")["Defect rates"].mean().reset_index()
)
print("--- Average Defect Rates by Supplier ---")
print(supplier_defects)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=supplier_defects,
    x="Supplier name",
    y="Defect rates",
    palette="rocket",
    hue="Supplier name",
    legend=False,
)
plt.title("Average Defect Rates by Supplier")
plt.show()
plt.savefig("supplier_defect_rates.png")
plt.close()
print(
    "\n[Success] Supplier defect rates printed and chart saved as 'supplier_defect_rates.png'."
)
```

```python

# Query 1: Customer Demographics & Purchasing Patterns

query_demo = """
    SELECT Customer_demographics, 
           Product_type,
           SUM(Number_of_products_sold) AS Total_Units_Sold,
           AVG(Revenue_generated) AS Avg_Revenue
    FROM supply_chain
    GROUP BY Customer_demographics, Product_type
    ORDER BY Total_Units_Sold DESC;
"""
result_demo = pd.read_sql_query(query_demo, conn)
print('--- 1. Customer Demographics by Product Type ---')
print(result_demo.head(5))
print('\n')
```

```python

# Query 2: Transportation Mode & Route Costs

query_transport = """
    SELECT Transportation_modes, 
           Routes,
           COUNT(SKU) AS Shipment_Count,
           AVG(Shipping_times) AS Avg_Shipping_Time,
           AVG(Costs) AS Avg_Total_Cost
    FROM supply_chain
    GROUP BY Transportation_modes, Routes
    ORDER BY Avg_Total_Cost DESC;
"""
result_transport = pd.read_sql_query(query_transport, conn)
print('--- 2. Transportation Modes & Route Costs ---')
print(result_transport.head(5))
print('\n')
```

```python

# Query 3: Supplier Quality Inspection & Defect Rates

query_inspection = """
    SELECT Supplier_name, 
           Location,
           Inspection_results,
           COUNT(SKU) AS SKU_Count,
           AVG(Defect_rates) AS Avg_Defect_Rate
    FROM supply_chain
    GROUP BY Supplier_name, Location, Inspection_results
    ORDER BY Avg_Defect_Rate DESC;
"""
result_inspection = pd.read_sql_query(query_inspection, conn)
print('--- 3. Supplier Inspection & Defect Audit ---')
print(result_inspection.head(5))
print('\n')
```

```python

# Query 4: Manufacturing Lead Time vs. Production Volume

query_production = """
    SELECT Supplier_name,
           Product_type,
           AVG(Manufacturing_lead_time) AS Avg_Mfg_Lead_Time,
           AVG(Production_volumes) AS Avg_Production_Volume,
           AVG(Manufacturing_costs) AS Avg_Mfg_Cost
    FROM supply_chain
    GROUP BY Supplier_name, Product_type
    ORDER BY Avg_Mfg_Lead_Time DESC;
"""
result_production = pd.read_sql_query(query_production, conn)
print('--- 4. Manufacturing Lead Time & Production Volumes ---')
print(result_production.head(5))

# Close the database connection
conn.close()
```
